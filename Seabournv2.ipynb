```python
import seaborn as sb
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Preplacementdata.csv')
```

```python
dataset
```

```python
# Number of rows and Columns 
print("The Number of Row:",dataset.shape[0])
print("The Number of Column",dataset.shape[1])
```

```python
sb.violinplot(
    x="gender",
    y="ssc_p",
    data=dataset,
    hue="gender",
    inner="quartile"
)

plt.show()


##sb.violinplot(x = "gender", y = "ssc_p", data = dataset)
#plt.show()
```

```python
percentiles = dataset.groupby("gender")["ssc_p"].quantile(
    [0.25, 0.50, 0.75]
).unstack()

percentiles.index.name = "Gender"
percentiles.columns = ["Q1", "Q2", "Q3"]

print(percentiles)
```

```python
#Please find the Observation on the above Violin chart:


Marks Scored
- Most male students scored between 60 and 65 marks.
- Most female students scored between 70 and 75 marks.
- Female students generally scored higher than male students.
Percentiles
- Q1: The 25th percentile.
- Q2: The 50th percentile (median).
- Q3: The 75th percentile.
- The thick black lines inside the violin plot show Q1, Q2, and Q3.
Conclusion
The median mark of female students is higher than that of male students. This indicates that female students performed better overall in SSC marks.

```
